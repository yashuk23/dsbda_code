```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import nltk

nltk.download('punkt')
nltk.download('stopwords')

sentence = "This is a sample sentence"

tokenized_sent = word_tokenize(sentence)
stop_words = set(stopwords.words("english"))
```

```python
filtered_sent = []

for w in tokenized_sent:
    if w not in stop_words:
        filtered_sent.append(w)

print("Tokenized Sentence:", tokenized_sent)
print("Filtered Sentence:", filtered_sent)
```

```python
from nltk.stem import PorterStemmer  
from nltk.tokenize import sent_tokenize, word_tokenize  
ps = PorterStemmer() 
stemmed_words=[]  
for w in filtered_sent:  
    stemmed_words.append(ps.stem(w))  
print("Filtered Sentence:",filtered_sent) 
print("Stemmed Sentence:",stemmed_words) 
```

```python
import pandas as pd 
import sklearn as sk 
import math 
```

```python
first_sentence = "Data Science is the best job of the 21st century" 
second_sentence = "machine learning is the key for d a science"
first_sentence= first_sentence.split(" ") 
second_sentence = second_sentence.split(" ")
total= set(first_sentence).union(set(second_sentence))
print(total) 
```

```python
wordDictA = dict.fromkeys(total, 0) 
wordDictB = dict.fromkeys(total, 0)
for word in first_sentence: 
    wordDictA[word]+=1 
for word in second_sentence: 
    wordDictB[word]+=1 
```

```python
pd.DataFrame([wordDictA, wordDictB]) 
```

```python
def computeTF(wordDict, doc):
    tfDict = {}
    corpusCount = len(doc)
    for word, count in wordDict.items():
        tfDict[word] = count / float(corpusCount)
    
    return tfDict
tfFirst = computeTF(wordDictA, first_sentence)
tfSecond = computeTF(wordDictB, second_sentence)

f = pd.DataFrame([tfFirst, tfSecond])
print(f)
```

```python
import math
first_sentence = ["this", "is", "a", "sample"]
second_sentence = ["this", "is", "another", "example"]

wordDictA = dict.fromkeys(set(first_sentence + second_sentence), 0)
wordDictB = dict.fromkeys(set(first_sentence + second_sentence), 0)

for word in first_sentence:
    wordDictA[word] += 1

for word in second_sentence:
    wordDictB[word] += 1

def computeIDF(docList):
    N = len(docList)
    idfDict = dict.fromkeys(docList[0].keys(), 0)

    for doc in docList:
        for word, val in doc.items():
            if val > 0:
                idfDict[word] += 1

    for word, val in idfDict.items():
        idfDict[word] = math.log10(N / float(val))

    return idfDict
idfs = computeIDF([wordDictA, wordDictB])
print(idfs)
```
