```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv("titanic.csv")   # use your file path
titanic.head()
```

```python
sns.countplot(x='survived', data=titanic)
plt.title("Survival Count")
plt.show()

# It shows:

# how many passengers survived
# how many did not survive
```

```python
sns.countplot(x='sex', hue='survived', data=titanic)
plt.title("Survival by Gender")
plt.show()
```

```python
sns.countplot(x='pclass', hue='survived', data=titanic)
plt.title("Survival by Class")
plt.show()
```

```python
# 🔹 4. Histogram for Fare Distribution
sns.histplot(titanic['fare'], bins=30, kde=True)
plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Frequency")
plt.show()

# A histogram is a graph used to show the distribution of numerical data.
```

```python
skew_value = titanic['fare'].skew()
print("Skewness of Fare:", skew_value)
```

```python
print("Mean Fare:", titanic['fare'].mean())
print("Median Fare:", titanic['fare'].median())
```
