```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag
from nltk.tokenize import sent_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')
nltk.download('punkt_tab')
```

```python
text = "Natural Language Processing is amazing and helps in analyzing text data easily."
```

```python
tokenize_text = sent_tokenize(text)
# Breaking sentence into words/tokens.
print(tokenize_text)
```

```python
tokens = word_tokenize(text) #sentence.split(" ")
print("Tokens:", tokens)
```

```python
# POS Tagging
pos = pos_tag(tokens)
# Assigns grammatical category to each word.
print("POS Tags:", pos)

# Tag	Meaning
# NN	Noun
# VB	Verb
# JJ	Adjective
# RB	Adverb
```

```python
# Stopword Removal
filtered = [word for word in tokens if word.lower() not in stopwords.words('english')]
print("After Stopword Removal:", filtered)
```

```python
# Stemming
ps = PorterStemmer()
stemmed = [ps.stem(word) for word in filtered]
print("Stemmed Words:", stemmed)

# Converts words to root form.
```

```python
# Lemmatization
lm = WordNetLemmatizer()
lemmatized = [lm.lemmatize(word) for word in filtered]
print("Lemmatized Words:", lemmatized)

# Converts words into meaningful dictionary words.
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
# Term Frequency -> How many times a word appears in a document.

documents = [
    "I love machine learning",
    "Machine learning is powerful"
]

# IDF ->How important or rare a word is across all documents.

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

print(tfidf_matrix.toarray())
# TF-IDF converts text into numeric importance scores.
```
