# data Normalization

```python
from sklearn import preprocessing
# only uses preprocessing module from sklearn
# Normalization->Mapping all Nominal value into Uniform Scale
    # Method->Min-Max Scaling ->Converts values between range 0 and 1.

    # Nor=X-Xmin/Xmax-Xmin
```

```python
x=df[['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']]
# Selects numerical columns from dataset.
```

```python
min_max_scaler=preprocessing.MinMaxScaler()
# Creates Min-Max normalization object.
```

```python
x_scaled=min_max_scaler.fit_transform(x)
# values become in range 0 to 1
```

```python
dataframe=pd.DataFrame(x_scaled)
```

```python
print(dataframe)
# Need of Normalization ->
# We do normalization because machine learning algorithms perform better when all features are on a similar scale.
```
