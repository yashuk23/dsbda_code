```python
import numpy as np
```

```python
import pandas as pd
```

```python
data=pd.read_csv("iris.csv")
```

```python
data.describe()
```

```python
data.SepalLengthCm.unique() 
# This returns all unique/distinct values present in the SepalLengthCm column.
```

```python
data.groupby(data.SepalLengthCm).count() 
# This groups the dataset by SepalLengthCm and counts the number of non-null values in each column for every group.
```

```python
data.groupby(data.SepalWidthCm).min() 
# This groups the dataset based on SepalWidthCm values and finds the minimum value of every column within each group.
```

```python
data.groupby(data.SepalWidthCm).max() 
# This groups the dataset based on SepalWidthCm values and finds the maximum value of every column within each group.
```

```python
data.groupby('SepalLengthCm').mean(numeric_only=True)
# This groups the dataset by SepalLengthCm and calculates the average (mean) of all numeric columns for each group.
```

```python
data.groupby('SepalLengthCm').std(numeric_only=True)
# This groups the dataset by SepalLengthCm and calculates the  Standard Deviation of all numeric columns for each group.
```
