```python
data = pd.read_csv("Social_Network_Ads.csv") 
data.head() 
```

```python
data.describe()
```

```python
plt.scatter(data['Age'], data['Purchased'],color='red') 
plt.xlabel("age") 
plt.ylabel("purchased") 
plt.show() 

X_train, X_test, y_train, y_test = train_test_split(data['Age'], data['Purchased'], 
test_size=0.20)
```

```python
from sklearn.linear_model import LogisticRegression 
```

```python
from sklearn.linear_model import LogisticRegression
import matplotlib.pyplot as plt1

model = LogisticRegression()
model.fit(X_train.values.reshape(-1, 1), y_train.values)

y_pred_sk = model.predict(X_test.values.reshape(-1, 1))

plt.clf()
plt.scatter(X_test, y_test, label="Actual")
plt.scatter(X_test, y_pred_sk, label="Predicted")

plt.legend()
plt.show()

print(f"Accuracy = {model.score(X_test.values.reshape(-1, 1), y_test.values)}")
```

```python
from sklearn.metrics import confusion_matrix

conf = confusion_matrix(y_test, y_pred_sk)

tn, fp, fn, tp = conf.ravel()
# ravel() ->Converts matrix into 1D array.

print("TN:", tn)
print("FP:", fp)
print("FN:", fn)
print("TP:", tp)

# A Confusion Matrix compares:

# actual values
# predicted values

# for classification problems.


# F1=2×Precision+Recall/Precision×Recall


# sigmoid() -> It converts any value into a probability between: 0 to 1 where x-input value
```

```python
accuracy = (tp + tn) / (tp + tn + fp + fn)
print("Accuracy =", accuracy)
```

```python
precision = tp / (tp + fp)
print("Precision {:0.2f}".format(precision))
```

```python
recall = tp / (tp + fn)
print("Recall {:0.2f}".format(recall))
```

```python
err = (fp + fn)/(tp + tn + fn + fp) 
print("Error rate {:0.2f}".format(err)) 
```
