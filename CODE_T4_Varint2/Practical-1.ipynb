```python
import pandas as pd
from sklearn import preprocessing
```

```python
iris = pd.read_csv("iris.csv")
```

```python
df = pd.DataFrame(iris,columns=["SepalLengthCm","SepalWidthCm","PetalLengthCm","PetalWidthCm"])
```

```python
df
```

```python
x=df[df.columns].values.astype(float)
```

```python
min_max_scalar=preprocessing.MinMaxScaler()
```

```python
x_scaled = min_max_scalar.fit_transform(x)
```

```python
df_normalized = pd.DataFrame(x_scaled)
```

```python
df_normalized
```

```python
df['Species']=iris['Species']
```

```python
df['Species'].unique()
```

```python
label_encoder=preprocessing.LabelEncoder()
```

```python
df['Species']=label_encoder.fit_transform(df['Species'])
```

```python
df['Species'].unique()
```

```python
df['Species'].dtype
```

```python
df
```

```python
features_df=df.drop(columns=['Species'])
```

```python
enc = preprocessing.OneHotEncoder()
```

```python
enc_df=pd.DataFrame(enc.fit_transform(df[['Species']]).toarray())
```

```python
df_encode = features_df.join(enc_df)
```

```python
df_encode
```

```python
df_encode.rename(columns={0:'Iris-Setosa',1:'Iris-Versicolar',2:'Iris-Verginica'},inplace=True)
```
